Merge BOTH datasets

```python
# check the cols name

print("Movies cols:")
print(movies.columns.tolist())

print("\nCredits cols:")
print(credits.columns.tolist())
```

```python
# merge both datasets - using movies titles - left join

movies = movies.merge(
    credits,
    left_on = "id",
    right_on = "movie_id",
    how = "left",
    suffixes=("", "_credits")
)

print("Merged datsets shape:",movies.shape)
print(movies.head())
```

```python
# Select the required columns

movies = movies[
    [
        "id",
        "title",
        "genres",
        "keywords",
        "overview",
        "cast",
        "crew"
    ]
].copy()

print("selected cols shape:", movies.shape)
```

```python
movies.head()
```

Handle missing values and duplicates

```python
# Remove duplicate movie records

movies = movies.drop_duplicates(subset="id")

# Fill missing text fields

text_columns = [
     "genres",
    "keywords",
    "overview",
    "cast",
    "crew"
]

movies[text_columns] = movies[text_columns].fillna("")

# Remove rows without a movie title
movies = movies.dropna(subset=["title"])

# Reset the index
movies = movies.reset_index(drop=True)

print("dataset shape after cleaning:", movies.shape)

print("missing vals:")
print(movies.isnull().sum())
```

Extract movie features

```python
# Create a function to extract names
import ast

def extract_names(text):
    try:
        data = ast.literal_eval(text)
        return [item["name"] for item in data]
    except (ValueError, SyntaxError, TypeError, KeyError):
        return []
```

```python
# Extract genres and keywords

movies["genres"] = movies["genres"].apply(extract_names)

movies["keywords"] = movies["keywords"].apply(extract_names)

movies[["title", "genres", "keywords"]].head()
```

```python
# Extract the top 3 cast members

def extract_cast(text):
    try:
        data = ast.literal_eval(text)
        return [item["name"] for item in data[:3]]
    except (ValueError, SyntaxError, TypeError, KeyError):
        return []
```

```python
movies["cast"] = movies["cast"].apply(extract_cast)

movies[["title", "cast"]].head()
```

```python
# Extract the director

def extract_director(text):
    try:
        data = ast.literal_eval(text)

        for person in data:
            if person.get("job") == "Director":
                return [person["name"]]

        return []

    except (ValueError, SyntaxError, TypeError, KeyError):
        return []
```

```python
movies["crew"] = movies["crew"].apply(extract_director)

movies.rename(columns={"crew": "director"}, inplace=True)

movies[["title", "director"]].head()
```

Clean the overview text

```python
movies["overview"] = (
     movies["overview"].apply(lambda text: text.split())
)
```

Check the extracted data

Combine movie features - tags

```python
# Combine all features into tags

movies["tags"] = (
    movies["overview"]
    + movies["genres"]
    + movies["keywords"]
    + movies["cast"]
    + movies["director"]
)

movies[["title", "tags"]].head()
```

```python
# Convert the tags list into text

movies["tags"] = movies["tags"].apply(
    lambda words: " ".join(words)
)

movies[["title", "tags"]].head()
```

```python
# Keep only the required columns

movies = movies[
    ["id", "title", "genres", "tags"]
].copy()

movies.head()
```

```python
# Normalize the text

movies["tags"] = movies["tags"].str.lower()

movies[["title", "tags"]].head()
```

```python
# Check the final dataset

print("Dataset shape:", movies.shape)
print("\nMissing values:")
print(movies.isnull().sum())

print("\nDuplicate movie IDs:", movies["id"].duplicated().sum())
print("\nExample movie tags:")
print(movies.loc[0, "tags"][:500])
```

Convert movie tags into numerical vectors

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features = 5000,
    stop_words = "english"
)

tfidf_matrix = tfidf.fit_transform(movies["tags"])

print("tfidf matrix shape:", tfidf_matrix.shape)
print("\n No.of movies:", tfidf_matrix[0])
print("No. of features:", tfidf_matrix[1])
```

Calculate movie similarity - score closer to 1 means greater similarity.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)

print("similarity matrix shape:", similarity.shape)
```

Create the recommendation function

```python
# Create a case-insensitive movie title lookup
indices = pd.Series(
    movies.index,
    index = movies["title"].str.lower()
).drop_duplicates()


def recommend_movies(movies_name, n=5):
  movies_name = movies_name.strip().lower()

  if movies_name not in indices:
    print(f"Movie '{movies_name} was not found.")
    print("Try entering a movie title from the dataset.")
    return

  movies_index = indices[movies_name]

  # Get similarity scores for the selected movie
  scores = list(enumerate(similarity[movies_index]))

  # Sort movies from most to least similar
  scores = sorted(
      scores,
      key = lambda x: x[1],
      reverse = True
  )

  # Skip the selected movie itself
  scores = scores[1: n+1]

  recommended_indices = [index for index, score in scores]

  recommendations = movies.iloc[recommended_indices][
      ["title", "genres"]
  ].copy()

  recommendations["similarity_score"] = [
      round(score, 3) for index, score in scores
  ]

  return recommendations.reset_index(drop=True)

```

Evaluate the recommendation system

```python
def evaluate_recommendations(movie_name, n=5):
    result = recommend_movies(movie_name, n)

    if result is None:
        print(f"'{movie_name}' not found.")
        return

    original_movie = movies[
        movies["title"].str.lower() == movie_name.lower()
    ].iloc[0]

    original_genres = set(original_movie["genres"])

    print("Movie:", movie_name)
    print("Original genres:", original_genres)
    print("\nRecommendations:")

    for _, row in result.iterrows():
        recommended_genres = set(row["genres"])

        common_genres = original_genres.intersection(
            recommended_genres
        )

        print(
            f"- {row['title']} | "
            f"Common genres: {len(common_genres)}"
        )
```

```python
evaluate_recommendations("The Dark Knight")
```

Calc quantitative evaluation score

```python
def genre_overlap_score(movie_name, n=5):
    result = recommend_movies(movie_name, n)

    if result is None:
        return None

    original_movie = movies[
        movies["title"].str.lower() == movie_name.lower()
    ].iloc[0]

    original_genres = set(original_movie["genres"])

    overlap_scores = []

    for _, row in result.iterrows():
        recommended_genres = set(row["genres"])

        common_genres = original_genres.intersection(
            recommended_genres
        )

        overlap = len(common_genres) / len(original_genres)

        overlap_scores.append(overlap)

    average_score = sum(overlap_scores) / len(overlap_scores)

    return average_score
```

```python
score = genre_overlap_score("The Dark Knight")

print(
    "Average Genre Overlap:",
    round(score * 100, 2),
    "%"
)
```

```python
test_movies = [
    "Avatar",
    "The Dark Knight",
    "Iron Man",
    "Titanic",
    "The Avengers"
]

for movie in test_movies:
    score = genre_overlap_score(movie)

    if score is not None:
        print(
            f"{movie}: {score * 100:.2f}%"
        )
```

Clean the final movies dataset

```python
# Keep only the columns actually needed by the project
movies = movies[
    ["id", "title", "genres", "tags"]
].copy()

# Final cleanup
movies["title"] = movies["title"].fillna("")
movies["genres"] = movies["genres"].apply(
    lambda x: x if isinstance(x, list) else []
)
movies["tags"] = movies["tags"].fillna("")

movies = movies.drop_duplicates(subset="id")
movies = movies.reset_index(drop=True)

print("Final dataset shape:", movies.shape)
print("\nColumns:")
print(movies.columns.tolist())

print("\nMissing values:")
print(movies.isnull().sum())
```

Rebuild the TF-IDF matrix

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

tfidf_matrix = tfidf.fit_transform(movies["tags"])

print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

Test the final model

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_movies(movie_name, n=5):

    movie_name = movie_name.strip().lower()

    indices = pd.Series(
        movies.index,
        index=movies["title"].str.lower()
    ).drop_duplicates()

    if movie_name not in indices:
        print("Movie not found.")
        return None

    movie_index = indices[movie_name]

    scores = cosine_similarity(
        tfidf_matrix[movie_index],
        tfidf_matrix
    ).flatten()

    scores[movie_index] = -1

    top_indices = scores.argsort()[::-1][:n]

    recommendations = movies.iloc[top_indices][
        ["title", "genres"]
    ].copy()

    return recommendations.reset_index(drop=True)
```
